```python
import cv2
import numpy as np
from ultralytics import YOLO
```

```python
# load model
model = YOLO('./models/yolo26n.pt')
```

```python
# Real-time prediction using the YOLO model
video = './videos/video-0.mp4'
cap = cv2.VideoCapture(video)

if not cap.isOpened():
    raise FileNotFoundError(f'Unable to open video: {video}')

try:
    while True:
        ret, frame = cap.read()
        if not ret:
            break

        results = model(frame, verbose=False)
        annotated_frame = results[0].plot() # output frame with numpy array
        #cv2.imshow('Video', frame)
        cv2.imshow('YOLO26 Annotated', annotated_frame)
        if cv2.waitKey(1) & 0xFF == ord('q'):
            break
finally:
    cap.release()
    cv2.destroyAllWindows()
```

### **Object Tracking**

```python
# Real-time prediction using the YOLO model
video = './videos/video-1.mp4'
cap = cv2.VideoCapture(video)

if not cap.isOpened():
    raise FileNotFoundError(f'Unable to open video: {video}')

try:
    while True:
        ret, frame = cap.read()
        if not ret:
            break

        results = model.track(frame, persist=True, verbose=False)
        annotated_frame = results[0].plot() # output frame with numpy array
        #cv2.imshow('Video', frame)
        cv2.imshow('YOLO26 Annotated', annotated_frame)
        if cv2.waitKey(1) & 0xFF == ord('q'):
            break
finally:
    cap.release()
    cv2.destroyAllWindows()
```
