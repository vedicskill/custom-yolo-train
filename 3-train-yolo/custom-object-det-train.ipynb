```python
import os
import torch
from ultralytics import YOLO
```

```python
# check the devices

if torch.cuda.is_available():
    device = 0
    print(torch.cuda.get_device_name(device))
else:
    device = "cpu"
    print("Using CPU")
```

```python
dataset = "dataset/data.yaml"
```

```python
# load the pretrain YOLO 26 model
model = YOLO("yolo26n.pt") # yolo26s, yolo26m, yolo26l, yolo26x
```

```python
# train model
results = model.train(
    data=dataset,
    device=device,
    epochs=100,
    imgsz=640,
    batch=16,
    workers=4,
    project="runs/yolo26",
    name="custom_yolo26",
    save=True,
    save_period=10,
    # validation
    val=True,
    patience=20,
    seed=42,
    cache=False,
    plots=True,
    verbose=True
)

# print the training success messages
print("Training completed successfully!")
print("Results:", results)
print("Saved model to:", "runs/yolo26/custom_yolo26")
# last model or checkpoint
print("Last model saved to:", "runs/yolo26/custom_yolo26/last.pt")
print("Best model saved to:", "runs/yolo26/custom_yolo26/best.pt")
```
