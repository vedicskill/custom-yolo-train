```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
# load the image
path = "./data/image-2.jpg"
img_cv = cv2.imread(path) # load image in BGR format
print(f"Original Image Shape: {img_cv.shape}")
```

```python
# display 
cv2.imshow("Original Image", img_cv)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

### Resize

```python
# Resize the image to a specific size
new_size = (448, 448) # width, height
img_resized = cv2.resize(img_cv, new_size)
print(f"Resized Image Shape: {img_resized.shape}")
```

```python
# display the image both original and resized
cv2.imshow("Original Image", img_cv)
cv2.imshow("Resized Image", img_resized)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# save resized image
cv2.imwrite("./data/image-2-resized.jpg", img_resized)
```

### Rotate

```python
# Rotate the image
# 90, 180 , 90 counter-clockwise(270 degrees clockwise)
img_rotated_90 = cv2.rotate(img_cv, cv2.ROTATE_90_CLOCKWISE)
img_rotated_180 = cv2.rotate(img_cv, cv2.ROTATE_180)
img_rotated_270 = cv2.rotate(img_cv, cv2.ROTATE_90_COUNTERCLOCKWISE) # 270 degrees clockwise


# display the rotated images
cv2.imshow("Original Image", img_cv)
cv2.imshow("Rotated 90 Degrees", img_rotated_90)
cv2.imshow("Rotated 180 Degrees", img_rotated_180)
cv2.imshow("Rotated 270 Degrees", img_rotated_270)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

### Convert Colour
https://docs.opencv.org/3.4.8/d8/d01/group__imgproc__color__conversions.html
- BGR2RGB
- BGR2RGBA
- BGR2GRAY
- RGB2BGRA

```python
# OpenCV uses BGR format by default, but Matplotlib uses RGB format. 
plt.imshow(img_cv)  # This will display the image in BGR format, which may look incorrect.
plt.title("Image Displayed using Matplotlib (Incorrect Color)")
plt.axis('off')
plt.show()
# So we need to convert the image from BGR to RGB before displaying it using Matplotlib.
img_rgb = cv2.cvtColor(img_cv, cv2.COLOR_BGR2RGB)
plt.imshow(img_rgb)
plt.title("Image Displayed using Matplotlib (Correct Color)")
plt.axis('off')
plt.show()
```

```python
# COLOR_BGR2RGBA
img_rgba = cv2.cvtColor(img_cv, cv2.COLOR_BGR2RGBA)
# 50% transparent
img_rgba[:, :, 3] = 128  # Set the alpha channel to 128 (50% transparency)
# display the RGBA image using Matplotlib
plt.imshow(img_rgba)
plt.title("Image Displayed using Matplotlib (RGBA with 50% Transparency)")
plt.axis('off')
plt.show()
```

```python
# BGR to Grayscale
img_gray = cv2.cvtColor(img_cv, cv2.COLOR_BGR2GRAY)
# look shape of the grayscale image
print(f"Original Image Shape: {img_cv.shape}")
print(f"Grayscale Image Shape: {img_gray.shape}")
```

```python
# display the image using matplotlib
plt.imshow(img_gray, cmap='gray')
plt.title("Grayscale Image Displayed using Matplotlib")
plt.axis('off')
plt.show()
```

### Smoothing Techniques

- Average Blur
- Gaussian Blur
- Bilateral Filter
- Median Blur

```python
img_cv = cv2.imread("./data/clock_wood.jpg") # load image in BGR format
# convert into RGB format
img_rgb = cv2.cvtColor(img_cv, cv2.COLOR_BGR2RGB)

plt.imshow(img_rgb)
plt.title("Clock Image Displayed using Matplotlib (Correct Color)")
plt.axis('off')
plt.show()
```

```python
# Average Blurring
img_blur = cv2.blur(img_cv, (5, 5)) # kernel size
img_blur_13 = cv2.blur(img_cv, (13,13)) # kernel size

# Subplot 1 x 3
plt.figure(figsize=(15, 5))
plt.subplot(1, 3, 1)
plt.imshow(img_rgb)
plt.title("Original Image")
plt.axis('off')

plt.subplot(1, 3, 2)
plt.imshow(img_blur)
plt.title("Blurred Image 5 x 5 Kernel")
plt.axis('off')

plt.subplot(1, 3, 3)
plt.imshow(img_blur_13) 
plt.title("Blurred Image 13 x 13 Kernel")
plt.axis('off')
plt.show()
```

```python
# Gaussian Blurring
img_gaussian_blur = cv2.GaussianBlur(img_cv, (5, 5), 0) # kernel size, sigmaX
img_gaussian_blur_13 = cv2.GaussianBlur(img_cv, (13, 13), 0) # kernel size, sigmaX
# as the kernel size increases, the image becomes more blurred
# display the images using matplotlib
plt.figure(figsize=(15, 5))
plt.subplot(1, 3, 1)
plt.imshow(img_rgb)
plt.title("Original Image")
plt.axis('off')

plt.subplot(1, 3, 2)
plt.imshow(img_gaussian_blur)
plt.title("Gaussian Blurred Image 5 x 5 Kernel")
plt.axis('off')

plt.subplot(1, 3, 3)
plt.imshow(img_gaussian_blur_13)
plt.title("Gaussian Blurred Image 13 x 13 Kernel")
plt.axis('off')
plt.show()
```

```python
# Bilateral Filtering
# Sharp edges are preserved while blurring the image. It is useful for noise reduction while keeping edges sharp.
img_bilateral = cv2.bilateralFilter(img_cv, 5, 75, 75) # diameter, sigmaColor, sigmaSpace
img_bilateral_13  = cv2.bilateralFilter(img_cv, 13, 75, 75) # diameter, sigmaColor, sigmaSpace

# display the images using matplotlib
plt.figure(figsize=(15, 5))
plt.subplot(1, 3, 1)
plt.imshow(img_rgb)
plt.title("Original Image")
plt.axis('off')

plt.subplot(1, 3, 2)
plt.imshow(img_bilateral)
plt.title("Bilateral Filtered Image 5 x 5 Kernel")
plt.axis('off')

plt.subplot(1, 3, 3)
plt.imshow(img_bilateral_13)
plt.title("Bilateral Filtered Image 13 x 13 Kernel")
plt.axis('off')
plt.show()
```

```python
cv2.imshow("Original Image", img_cv)
cv2.imshow("Gaussian Blurred Image", img_gaussian_blur)
cv2.imshow("Bilateral Filtered Image", img_bilateral_13)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

#### Median Blur
- Use to remove salt pepper type of noise in the image

```python
img_sp = cv2.imread("./data/sp_noise.jpg") # load image in BGR format
img_sp_rgb = cv2.cvtColor(img_sp, cv2.COLOR_BGR2RGB) # convert BGR to RGB

plt.imshow(img_sp_rgb)
plt.title("Salt and Pepper Noise")
plt.axis('off')
plt.show()
```

```python
# Median Blurring
img_median_blur = cv2.medianBlur(img_sp_rgb, 3) # kernel size

# create the subplot 1 x 2
plt.figure(figsize=(15, 5))
plt.subplot(1, 2, 1)
plt.imshow(img_sp_rgb)
plt.title("Salt and Pepper Noise")
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(img_median_blur)
plt.title("Median Blurred")
plt.axis('off')
plt.show()
```

```python
# save the median blurred image
cv2.imwrite("./data/sp_noise_median_blur.jpg", cv2.cvtColor(img_median_blur, cv2.COLOR_RGB2BGR))
```

### Image Thresholding

```python
img_cv = cv2.imread("./data/image-4.jpg", cv2.IMREAD_GRAYSCALE) # load image in grayscale format

plt.imshow(img_cv, cmap='gray')
plt.title("Grayscale Image Displayed using Matplotlib")
plt.axis('off')
plt.show()
```

```python
# Threshold Binary
# Thresholding is a technique used to segment an image into different regions based on pixel intensity values
# if pixel value > threshold value, it is set to max value (255) [White], else it is set to 0 (Black)
# cv2.THRESH_BINARY: if pixel value > threshold value, it is set to max value (255) [White], else it is set to 0 (Black)
threshold_value = 50
max_value = 255
ret, thresh_binary = cv2.threshold(img_cv, threshold_value, max_value, cv2.THRESH_BINARY) # threshold value, max value, threshold type
ret, thresh_binary_inv = cv2.threshold(img_cv, threshold_value, max_value, cv2.THRESH_BINARY_INV) # threshold value, max value, threshold type

plt.figure(figsize=(15, 5))
plt.subplot(1, 2, 1)
plt.imshow(thresh_binary, cmap='gray')
plt.title("Threshold Binary Image Displayed using Matplotlib")  
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(thresh_binary_inv, cmap='gray')
plt.title("Threshold Binary Inverse Image Displayed using Matplotlib")
plt.axis('off')

plt.show()
```

```python
# Adaptive Thresholding
# Adaptive thresholding is a technique used to segment an image into different regions based on pixel intensity
img_blur = cv2.GaussianBlur(img_cv, (9, 9), 0) # kernel size, sigmaX
img_bilateral = cv2.bilateralFilter(img_cv, 9, 50, 30) # d, sigmaColor, sigmaSpace
adaptive_threshold = cv2.adaptiveThreshold(img_cv, 255, cv2.ADAPTIVE_THRESH_MEAN_C, cv2.THRESH_BINARY, 11, 2)
adaptive_threshold_blur = cv2.adaptiveThreshold(img_blur, 255, cv2.ADAPTIVE_THRESH_MEAN_C, cv2.THRESH_BINARY, 11, 2)
adaptive_threshold_gaussian = cv2.adaptiveThreshold(img_blur, 255, cv2.ADAPTIVE_THRESH_GAUSSIAN_C, cv2.THRESH_BINARY, 11, 2)
adaptive_threshold_bilateral = cv2.adaptiveThreshold(img_bilateral, 255, cv2.ADAPTIVE_THRESH_GAUSSIAN_C, cv2.THRESH_BINARY, 11, 2)

plt.figure(figsize=(15, 5))
plt.subplot(1, 2, 1)
plt.imshow(adaptive_threshold, cmap='gray')
plt.title("Adaptive Thresholding Image Displayed using Matplotlib")
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(adaptive_threshold_bilateral, cmap='gray')
plt.title("Adaptive Thresholding with Bilateral Filtered Image")
plt.axis('off')

plt.show()
```
