```python
import numpy as np
import cv2 # opencv
from PIL import Image # Python image Library
from matplotlib import image as mpimg # matplotlib image
```

```python
img_path = "./data/image-1.jpg"

# Read image using OpenCV
img_cv2 = cv2.imread(img_path) # opencv will return none if the image is not found
if img_cv2 is None:
    print("Error: Could not read the image.")

print(img_cv2.shape) # (height, width, channels - Blue, Green, Red)
```

```python
# display image using OpenCV
cv2.imshow("Image using OpenCV", img_cv2)
# Wait for a key press and close the image window
cv2.waitKey(0) # 0 represents wait indefinitely until a key is pressed (ESC, Enter, Space, etc.)
# 5000 represents wait for 5000 milliseconds (5 seconds) before closing the window
cv2.destroyAllWindows() # Close all OpenCV windows
```

```python
# Save the image using OpenCV
output_path = "./data/output-image.png"
cv2.imwrite(output_path, img_cv2)
```

```python
# Read image using PIL
img_pil = Image.open(img_path)

# display image using PIL
img_pil.show() # This will open the default image viewer on your system
```

```python
# Save the image using PIL
output_path_pil = "./data/output-image-pil.png"
img_pil.save(output_path_pil)
```

```python
# Read image using Matplotlib
img_matplotlib = mpimg.imread(img_path) # RGB not BGR(OpenCV)
print(img_matplotlib.shape) # (height, width, channels - Red, Green, Blue)
# display image using Matplotlib
import matplotlib.pyplot as plt
plt.imshow(img_matplotlib)
plt.axis("off")
plt.show()
```

```python
# save the image using Matplotlib
output_path_matplotlib = "./data/output-image-matplotlib.png"
plt.imsave(output_path_matplotlib, img_matplotlib)
```
