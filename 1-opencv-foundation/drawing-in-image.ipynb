```python
import cv2
import numpy as np
from matplotlib import pyplot as plt
```

```python
img = np.zeros((1000, 1500, 3), np.uint8)
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

plt.imshow(img)
plt.show()
```

```python
# draw a straight line from point (100, 100) to point (500, 500) with color red and thickness 5
# parameters for line: image, start_point, end_point, color, thickness
cv2.line(img, (350, 200), (500, 500), (255, 0, 0), 5)

# parameters for putText: image, text, position, font, fontScale, color, thickness
# fontScale is the size of the font in pixels, color is in BGR format, thickness is the thickness of the text
cv2.putText(img, '(350, 200)', (350, 200), cv2.FONT_HERSHEY_SIMPLEX, 1, (255, 255, 255), 2) 
cv2.putText(img, '(500, 500)', (500, 500), cv2.FONT_HERSHEY_SIMPLEX, 1, (255, 255, 255), 2)



plt.imshow(img)
plt.xlabel('X →')
plt.ylabel('← Y')
plt.show()
```

```python
# draw rectangle

cv2.rectangle(img, (350, 200), (500, 500), (0, 255, 0), 5)

# parameters for putText: image, text, position, font, fontScale, color, thickness
# fontScale is the size of the font in pixels, color is in BGR format, thickness is the thickness of the text
cv2.putText(img, '(350, 200)', (350, 200), cv2.FONT_HERSHEY_SIMPLEX, 1, (255, 255, 255), 2) 
cv2.putText(img, '(500, 500)', (500, 500), cv2.FONT_HERSHEY_SIMPLEX, 1, (255, 255, 255), 2)



plt.imshow(img)
plt.xlabel('X →')
plt.ylabel('← Y')
plt.show()
```

```python
# draw circle

cv2.circle(img, (1000,800),500,(0,255,255),5)
cv2.circle(img, (1000,800),300,(0,0,255),5)
cv2.circle(img, (1000,800),200,(255,0,255),5)

# parameters for putText: image, text, position, font, fontScale, color, thickness
# fontScale is the size of the font in pixels, color is in BGR format, thickness is the thickness of the text
cv2.putText(img, '(350, 200)', (350, 200), cv2.FONT_HERSHEY_SIMPLEX, 1, (255, 255, 255), 2) 
cv2.putText(img, '(500, 500)', (500, 500), cv2.FONT_HERSHEY_SIMPLEX, 1, (255, 255, 255), 2)



plt.imshow(img)
plt.xlabel('X →')
plt.ylabel('← Y')
plt.show()
```
