```python
import numpy as np
import cv2
import matplotlib.pyplot as plt
```

```python
# create an image
img = np.zeros((1000,1500,3),dtype=np.uint8)
# convert this image to RGB
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

plt.imshow(img)
plt.show()
```

```python
# DRAW A LINE
pt1 = (100,200)
pt2 = (500,500)
color = (255, 0 , 0) # R, G, B
thickness = 5
cv2.line(img, pt1, pt2, color, thickness)
cv2.line(img, (100, 150), pt2, (0, 255, 0), 10)
cv2.line(img, (100, 100), pt2, (0, 0, 255), 20)

# put text
cv2.putText(img, '(100, 100)', (100,100) , cv2.FONT_HERSHEY_SIMPLEX, 1, (255, 255, 255), 2)
cv2.putText(img, '(500, 500)', (500,500) , cv2.FONT_HERSHEY_SIMPLEX, 3, (255, 255, 255), 2)

plt.imshow(img)
plt.xlabel('X →')
plt.ylabel('← Y')
plt.show()
```

```python
# DRAW A LINE
pt1 = (100,200)
pt2 = (500,500)
color = (255, 0 , 0) # R, G, B
thickness = 5
cv2.line(img, pt1, pt2, color, thickness)
cv2.line(img, (100, 150), pt2, (0, 255, 0), 10)
cv2.line(img, (100, 100), pt2, (0, 0, 255), 20)
# Rectangle
cv2.rectangle(img, (100, 100), pt2, (255, 255, 0), 10)

# put text
cv2.putText(img, '(100, 100)', (100,100) , cv2.FONT_HERSHEY_SIMPLEX, 1, (255, 255, 255), 2)
cv2.putText(img, '(500, 500)', (500,500) , cv2.FONT_HERSHEY_SIMPLEX, 3, (255, 255, 255), 2)

plt.imshow(img)
plt.xlabel('X →')
plt.ylabel('← Y')
plt.show()
```
