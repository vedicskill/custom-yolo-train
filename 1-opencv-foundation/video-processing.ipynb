```python
import cv2
import numpy as np
import time
```

```python
video_path = './data/birds-video.mp4'

cap = cv2.VideoCapture(video_path) # video_path, 0 = default web camera, 1 = external camera

while True:
    ret, frame = cap.read()
    # ret = true, read video sucessfully 
    # ret == false, failed to read video

    if ret is not True: 
        print("Failed to read Video")
        break

    # Apply bilateral filter for noise reduction while keeping edges sharp
    # blur = cv2.bilateralFilter(frame, 9, 50, 30)
    # Apply Gaussian Blur for further noise reduction
    blur = cv2.GaussianBlur(frame, (3, 3), 0)
    # convert the frame to grayscale
    gray = cv2.cvtColor(blur, cv2.COLOR_BGR2GRAY)
    # Apply Canny Edge Detection
    edge = cv2.Canny(gray, 100, 200)
    
    cv2.imshow("Video Capture",frame)
    cv2.imshow("Grayscale Video", gray)
    cv2.imshow("Edge Detection", edge)
    # fps = 1000 / (no of frames)
    if cv2.waitKey(int(1000/24)) == ord('q'):
        break

cap.release()
cv2.destroyAllWindows()
```
