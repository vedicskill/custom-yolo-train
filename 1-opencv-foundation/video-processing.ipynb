```python
import cv2
import numpy as np
import time
```

```python
video_path = './data/birds-video.mp4'

cap = cv2.VideoCapture(video_path) # video_path, 0 = default web camera, 1 = external camera

while True:
    ret, frame = cap.read()
    # ret = true, read video sucessfully 
    # ret == false, failed to read video

    if ret is not True: 
        print("Failed to read Video")
        break

    cv2.imshow("Video Capture",frame)

    # fps = 1000 / (no of frames)
    if cv2.waitKey(int(1000/24)) == ord('q'):
        break

cap.release()
cv2.destroyAllWindows()
```
